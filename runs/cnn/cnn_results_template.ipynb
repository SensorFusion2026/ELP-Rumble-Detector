# CNN Results

Display-only notebook for showcasing one CNN training run.
Figures are pre-generated by evaluate_cnn.py; this notebook only loads metrics and images.

```python
import json
from pathlib import Path

import pandas as pd
from IPython.display import Image, display

RUN_DIR = Path(".")
FIGURES_DIR = Path("figures")
```

## Test Metrics Summary

```python
metrics_path = RUN_DIR / "test_metrics.json"
if not metrics_path.exists():
    raise FileNotFoundError(f"Expected metrics file not found: {metrics_path}")

with open(metrics_path) as f:
    metrics = json.load(f)

skip = {"confusion_matrix"}
rows = [(k, v) for k, v in metrics.items() if k not in skip]
metrics_df = pd.DataFrame(rows, columns=["Metric", "Value"]).set_index("Metric")
display(metrics_df)
```
